# Advantages of Machine Learning

Automation: ML removes the need for constant human supervision in routine tasks (e.g., spam filtering, sorting emails), significantly boosting productivity.

Pattern Recognition: It can rapidly sift through massive amounts of data to find hidden trends that are completely invisible to humans (e.g., predicting consumer behavior or diagnosing rare diseases).

Continuous Improvement: As ML algorithms process more experience and real-time data, they increase in accuracy and efficiency over time.

Personalization: Enhances user experiences by tailoring recommendations, content, and advertisements to individual preferences (like those used on Netflix or Amazon).

# Disadvantages of Machine Learning

Massive Data Dependency: ML algorithms require massive amounts of clean, diverse, and well-labeled data to function correctly. Without it, the models perform poorly.

High Computational Costs: Advanced modeling techniques, like deep learning, demand significant computing power and specialized infrastructure, which can be expensive to build and maintain.

Bias and Error: If the training data is skewed or unrepresentative, the model will produce biased or flawed predictions, which can sometimes go undetected for long periods.

The "Black Box" Effect: It can be very difficult for humans to interpret how and why an algorithm arrived at a specific decision or prediction, making debugging and troubleshooting a challenge

```python
from sklearn.linear_model import LinearRegression
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
data =  {
    "exp" : [1,2,3,4,5,6,7,8],
    "Salary": [200,300,100,2200,11000,100000,90000,500000]
}
```

```python
df = pd.DataFrame(data)
```

```python
y = df[['exp']]
x = df[['Salary']]
```

```python
model = LinearRegression()
```

```python
model.fit(x,y)
```

```python
salary = model.predict([[10000]])
print('pred exp: ', salary)
```

```python
data = {
    "Month" : [1,2,3,4,5,6],
    "Demand":[500,550,600,650,700,750]
}
```

```python
df = pd.DataFrame(data)
```

```python
x = df[['Month']]
y = df[['Demand']]
```

```python
data = {
    "adv_bud" : [100,300,500,600,800,900],
    "sales" : [100,400,700,800,1000,1300]
}

df = pd.DataFrame(data)

y = df[['sales']]
x = df[['adv_bud']]

model = LinearRegression()
model.fit(x,y)

sales = model.predict([[800]])
print("sales pred: " , sales)
```

```python

data = {
    "income" : [10000,50000,60000,70000,20000,30000,33000,40000,49990],
    "status" : [0,1,1,1,0,0,0,0,0]
}

df = pd.DataFrame(data)

model = DecisionTreeClassifier()

x = df[['income']]
y = df[['status']]

model.fit(x,y)

status = model.predict([[49999]])

print("pred loan status: not approved" if not status  else "pred loan status: approved")
```
