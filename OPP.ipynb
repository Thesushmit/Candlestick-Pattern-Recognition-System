METHOD

Method is a function inside class 

method defines behavior of object

```python
class car:
    def __init__(self, model , price , company):
        self.model = model 
        self.price = price 
        self.company = company
    
    def modelS(self):
        return f"car name: {self.model} \ncar price:  {self.price}\ncar company: {self.company}"
    
    def priceonroad(self):
        tax = 1000
        return tax + self.price
    
    def __str__(self):
        return f"{self.model} on road : {self.priceonroad()}"
    
    
        
c1 = car('endevor', 200000, 'ford' )
```

```python
print(c1.modelS())
```

```python
print(c1)
```

ENCAPUSLATION

Binding data (variable) and Methods (function) together in a class 



```python
class airplane:
    def __init__(self, no_airplane , type, name):
        self.no_airplane = no_airplane
        self.type = type 
        self.name = name
        
    def display(self):
        print(f"Name : {self.name}\nType: {self.type}\nno_airplane: {self.no_airplane}")        
```

Note 

Name and age are data

display() function or method

both are wrapped inside a class student

```python
airplane1 = airplane(100, "domestic", "AirIndia")
airplane1.display()
```

POLYMORPHISM



```python
class circle:
    def area(self):
        print("Area of circle: (22/7)^2r")
class rectangle:
    def area(self):
        print("Area of rectangle: l*w")
class square: 
    def area(self):
        print("Area of square: side*side")
        
c = circle()
r = rectangle()
s = square()

c.area()
r.area()
s.area()
```

INHERITANCE 

Acquiring properties from parent class to child class 

```python
#syntax 

class parent:
    pass
class child(parent):
    pass
```

SINGLE LEVEL INHERITANCE

```python
class parent :
    def name(self):
        return "Ram" 
class child(parent):
    def display(self):
        print(f"my father name is {self.name()}")
        
c1 = child()
c1.display()
```

MULTIPLE INHERITANCE 

One Child Inherites multiple parents class

```python
class parent2: 
    def name1(self):
        return "Rita"
class parent1 :
    def name(self):
        return "Rama" 
class child(parent1, parent2):
    def display(self):
        print(f"my father name is {self.name()} and my mother name is {self.name1()}")
        
c1 = child()
c1.display()
```

MULTILEVEL Inheritance

```python
class dadaji:
    def name2(self):
        return "dadaji"
class parent2(dadaji): 
    def name1(self):
        return "Rita"
class parent1(dadaji):
    def name(self):
        return "Rama" 
class child(parent1, parent2):
    def display(self):
        print(f"my father name is {self.name()} and my mother name is {self.name1()},and my grandfather name is {self.name2()}")
        
c1= child()
c1.display()
```

HIERARCHICAL INHERITANCE - one parent multiple childs

```python
class dadaji:
    def name2(self):
        return "dadaji"
class parent2(dadaji): 
    def name1(self):
        return "Rita"
class parent1(dadaji):
    def name(self):
        return "sushmit" 
class child(parent1, parent2):
    def display(self):
        print(f"my father name is {self.name()} and my mother name is {self.name1()},and my grandfather name is {self.name2()}")
        
c1= child()
c1.display()
```

HYBRID Inheritance 

```python
class dadaji(child):
    def name2(self):
        return "dadaji"
class parent2(dadaji): 
    def name1(self):
        return "Rita"
class parent1(dadaji):
    def name(self):
        return "sushmit" 
class child(parent1, parent2):
    def display(self):
        print(f"my father name is {self.name()} and my mother name is {self.name1()},and my grandfather name is {self.name2()}")
        
c1= dadaji()
c1.display()
```
