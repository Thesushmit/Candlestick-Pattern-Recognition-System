```python
import numpy as np
```

array verison

```python
print(np.__version__)
```

multi dimension array

```python
# 0-D array
print(np.array(12))
```

```python
d = np.array([[1,2 ],[2,2]])
print(np.ndim(d))
```

```python
# higher dimension array

d = np.array([1,1,2], ndmin=4)
print(d)
```

array slicing

```python
ar = np.array([1,2,3])
print(ar[0])
```

```python
ar = np.array(
            [
                [
                    [1,2,3,4],
                    [5,6,7,8]
                ]
            ]
        )
print(np.ndim(ar))
print(ar[0,0,1:4])

# ar1 = np.array([[[0,1,1],[0,102,1]],[[1,2],[12]]])
```

array datatype and typecasting

```python
arr = np.array([0,2,4,5,6])
print(arr.dtype)

arr_new = np.array([1,3,5], dtype='f')
print(arr_new)
```

```python
arr = arr.astype('U')
print(arr)
```

array copy and view 

```python
new = arr.copy()
arr[0] = 'BMW'
print("new: ",new)
print("arr: ",arr)
```

```python
vnew = arr.view()
arr[1] = 'Mercedies'
print("vnew:",vnew)
print("arr:",arr)
```
