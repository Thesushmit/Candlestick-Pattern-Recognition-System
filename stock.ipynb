```python
def hammer(open , high , low , close): 
    body = abs(open-close)
    upperwrick = high-max(close, open)
    lowerwrick = min(close, open) - low
    
    if lowerwrick >= 2 * body and upperwrick <= body:
        return True
    return False

def shootingstar(open , high , low , close):
    body = abs(open-close)
    upperwrick = high- max(open, close)
    lowerwrick = min(close,open) - low
    
    if upperwrick >= 2*body and lowerwrick <= body:
        return True
    return False

def bullish_engulfing(curr_open , curr_close, prev_open , prev_close):
    if curr_open < curr_close and prev_open > prev_close and curr_open <= prev_close and curr_close >= prev_open:
        return True 
    return False

def bearish_engulfing(curr_open , curr_close, prev_open , prev_close):
    if curr_open > curr_close and prev_close > prev_open and curr_open >= prev_close and curr_close <= prev_open:
        return True
    return False

i = 0
while(i < len(data)-1):
    if hammer(data.loc[i,'Open'], data.loc[i, 'High'], data.loc[i, 'Low'], data.loc[i,'Close']):
        data.loc[i,'Hammer'] = True
    if shootingstar(data.loc[i,'Open'], data.loc[i, 'High'], data.loc[i, 'Low'], data.loc[i,'Close']):
        data.loc[i,'Shooting_star'] = True
    if bullish_engulfing(data.loc[i+1,'Open'], data.loc[i+1,'Close'],data.loc[i,'Open'], data.loc[i,'Close']):
        data.loc[i+1,'Bullish_engulfing'] = True
    if bearish_engulfing(data.loc[i+1,'Open'], data.loc[i+1,'Close'],data.loc[i,'Open'], data.loc[i,'Close']):
        data.loc[i+1,'Bearish_engulfing'] = True
        
    i = i + 1
```
